```python
#1. _init_client()

def _init_client(self):
    if self.persist_dir:
        return chromadb.PersistentClient(path=self.persist_dir)

    return chromadb.Client()
#Có persist_dir → persistent mode
#Không có persist_dir → in-memory mode
```

```python
#2. add()
def add(self, chunks, embeddings):
    if len(chunks) != len(embeddings):
        raise ValueError(
            "chunks and embeddings must have same length"
        )

    ids = [f"chunk_{i}" for i in range(len(chunks))]

    self.collection.add(
        ids=ids,
        documents=chunks,
        embeddings=embeddings
    )

    return True

#Nhận chunks và embeddings
#Lưu vào Chroma
#Trả về True
```

```python
#3. get_collection_stats()

def get_collection_stats(self):
    return {
        "collection_name": self.collection_name,
        "count": self.collection.count()
    }
#Trả về thống kê collection
#Ít nhất phải có số lượng document/vector
```

```python
#4. delete_collection()

def delete_collection(self):
    self.client.delete_collection(
        name=self.collection_name
    )
    return True
#Xóa toàn bộ collection
#Trả về True
```
